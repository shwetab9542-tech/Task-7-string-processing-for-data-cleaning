```python
#Task: String Processing for Data Cleaning
#1. Import Pandas
import pandas as pd
```

```python
#2. Basic String Methods
text = "  Hello Data Science  "

print("Original:", text)
print("strip():", text.strip())
print("lower():", text.lower())
print("upper():", text.upper())
print("replace():", text.replace("Data", "Python"))
print("split():", text.split())
```

```python
#3. Cleaning Names
names = ["  Rahul Sharma", "PRIYA PATIL  ", " amit joshi ", "Sneha   Deshmukh"]

clean_names = []

for name in names:
    name = name.strip()
    name = " ".join(name.split())
    name = name.title()
    clean_names.append(name)

print(clean_names)
```

```python
#4. Cleaning Email Addresse
emails = [
    "  RAHUL@GMAIL.COM ",
    "priya@gmail.com ",
    " AMIT@YAHOO.COM",
    "sneha @gmail.com"
]

clean_emails = []

for email in emails:
    email = email.strip()
    email = email.lower()
    email = email.replace(" ", "")
    clean_emails.append(email)

print(clean_emails)
```

```python
#5. Cleaning Categories
categories = [" Male", "FEMALE", "male ", " Female ", "MALE"]

clean_categories = []

for category in categories:
    category = category.strip().lower()
    category = category.title()
    clean_categories.append(category)

print(clean_categories)
```

```python
#6. Using split()
full_name = "Shweta Borade"

parts = full_name.split()

print("First Name:", parts[0])
print("Last Name:", parts[1])
```

```python
#7. Checking Unwanted Spaces
names = ["Rahul", " Priya", "Amit ", " Sneha "]

for name in names:
    if name != name.strip():
        print("Unwanted spaces found:", repr(name))
    else:
        print("No unwanted spaces:", repr(name))
```

```python
#8. Small Text-Cleaning Function
def clean_text(text):
    text = text.strip()
    text = " ".join(text.split())
    text = text.lower()
    return text

text = "   Hello   Data   Science   "

print("Original:", text)
print("Cleaned:", clean_text(text))
```

```python
#9. Customer Dataset Example
data = {
    "Name": ["  Rahul Sharma", "PRIYA PATIL ", " amit joshi"],
    "Email": [" RAHUL@GMAIL.COM ", "Priya@Gmail.com", " AMIT@YAHOO.COM "],
    "Category": [" customer", "CUSTOMER ", " Customer"]
}

df = pd.DataFrame(data)

print("Original Data:")
display(df)
```

```python
#10. Apply String Cleaning
df["Name"] = df["Name"].str.strip().str.title()

df["Email"] = df["Email"].str.strip().str.lower().str.replace(" ", "", regex=False)

df["Category"] = df["Category"].str.strip().str.lower().str.title()

print("Cleaned Data:")
display(df)
```

```python
#11. Final Result
print("Final Cleaned Dataset:")
display(df)
```
